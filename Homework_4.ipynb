```python
#a7
def prime(i):
    var = True
    if i <2:
        var= False
    else:
        for j in range(2,int(i**0.5)+1):
            if i%j ==0:
                var = False
    return var

try:
    n = int(input('Nhập số nguyên dương n: '))
    M = 1
    for j in range(n,1,-1):
        if n%j == 0 and prime(j):
            if j>M:
                M = j
    print(f'ước nguyên tố lớn nhất là: {M}')


except ValueError: 
    print('Vui lòng nhập đúng số hợp lệ')
    
```

```python
#a8
def palindrome(n):
    n = str(n)
    if n != n[::-1]:
        var = False
    else:
        var = True
    return var

try:
    n = int(input('Nhập số nguyên dương n: '))
    count = 0
    while n <0:
        n = int(input('Nhập số nguyên dương n: '))
    else:
        while not palindrome(n):
            n += int(str(n)[::-1])
            count += 1
        print(n,count)
        
except ValueError: 
    print('Vui lòng nhập đúng số hợp lệ')

```

```python
def rever(n):
    v = 0
    while n != 0:
        v = v*10 + n%10
        n //=10
    return v
def palin(n):
    n = str(n)
    if n == n[::-1]:
        var = True
    else:
        var = False
    return var
        
n = int(input())
count = 0
while not palin(n):
    n += rever(n)
    count += 1
print(n)
print(count)
    
```

```python
#a10
n = int(input('Nhập số nguyên dương: '))
def collatz(n):
    count = 0
    while n !=1:
        count +=1
        if n%2 ==0:
            n =n/2
        else:
            n = 3*n+1
    return count
M = 1
for i in range (1,n+1):
    if collatz(i)>collatz(M):
        M = i
    elif collatz(i)==collatz(M):
        M = min(i,M)
print(f'({M} {collatz(M)})')
```
