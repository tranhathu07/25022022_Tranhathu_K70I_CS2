```python
#a7
def prime(i):
    var = True
    if i <2:
        var= False
    else:
        for j in range(2,int(i**0.5)+1):
            if i%j ==0:
                var = False
    return var

try:
    n = int(input('Nhập số nguyên dương n: '))
    M = 1
    for j in range(n,1,-1):
        if n%j == 0 and prime(j):
            if j>M:
                M = j
    print(f'ước nguyên tố lớn nhất là: {M}')


except ValueError: 
    print('Vui lòng nhập đúng số hợp lệ')
    
```

```python
#a8
def palindrome(n):
    n = str(n)
    var = True
    for i in range(len(n)//2+1):
        if n[i] != n[-i-1]:
            var = False
            break
    return var

try:
    n = int(input('Nhập số nguyên dương n: '))
    count = 0
    while n <0:
        n = int(input('Nhập số nguyên dương n: '))
    else:
        while not palindrome(n):
            n += int(str(n)[::-1])
            count += 1
        print(n,count)
        
except ValueError: 
    print('Vui lòng nhập đúng số hợp lệ')

```

```python
#a9
try:
    n = int(input('Nhập số nguyên dương n: '))
    while n <0:
        n = int(input('Nhập số nguyên dương n: '))
    else:
        ds =[]
        for i in range(1,n+1):
            for j in range(1,int(i**0.5)+1):
                if j*j ==i:
                    ds.append(i)
    if ds:
        print(' '.join(str(x) for x in ds))
    if not ds:
        print('No number')
except ValueError: 
    print('Vui lòng nhập đúng số hợp lệ')
```

```python
#a10
n = int(input('Nhập số nguyên dương: '))
def collatz(n):
    count = 0
    while n !=1:
        count +=1
        if n%2 ==0:
            n =n/2
        else:
            n = 3*n+1
    return count
M = 1
for i in range (1,n+1):
    if collatz(i)>collatz(M):
        M = i
    elif collatz(i)==collatz(M):
        M = min(i,M)
print(f'({M} {collatz(M)})')
```
