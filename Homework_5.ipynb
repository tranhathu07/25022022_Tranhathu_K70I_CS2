```python
#W5A3: 
ds = [True] * 10**6
ds[0] = False
ds[1] = False
for i in range(2, 10**3):
    for j in range(i*i,10**6,i):
        ds[j] = False

def prime(n):
    if ds[n]:
        print('Đây là số nguyên tố')
    else:
        print('Ko phải số nguyên tố')
n = int(input())
prime(n)
```

```python
#W5A4
n = int(input())
def shh(n):
    uoc =[1]
    for i in range(2,int(n**(1/2))+1):
        if n % i == 0:
            uoc.append(i)
            uoc.append(n/i)
    if sum(uoc) == n:
        return True
    else:
        return False
if shh(n):
    print('so hoan hao')
else:
    print('so ko hoan hao')
```

```python
#W5A5
lst = list(map(int,input().split()))
k = int(input())
def tim(lst,k):
    found = -1
    for i in range(len(lst)):
        if lst[i] == k:
            found = i
            break
    return found
tim(lst,k)
    
```

```python
#W5A6
n = int(input())
def giaithua(n):
    if n<0:
        print('Vui long nhap lai')
        
    if n == 0:
        return 1
    if n == 1:
        return 1
    if n>1:
        return n* giaithua(n-1)
giaithua(n)
```

```python
n1,n2,op = input().split()
def casio(n1,n2,op):
    n1,n2 = float(n1),float(n2)
    if op == '-':
        return n1 - n2
    if op == '+':
        return n1+n2
    if op =='*':
        return n1*n2
    if op == '/':
        return n1/n2
casio(n1,n2,op)
        
    
```

```python
#W5A8
x,y = map(int,input().split())
def hamming(x,y):
    return bin(x^y).count('1')
hamming(x,y)
```

```python
#W5A9
n = int(input())
def tong(n):
    su = 0
    while n != 0:
        r = n% 10
        su += r
        n //= 10
    return su
tong(n)
```

```python
#W5A10
a = input()
b = input()
def dangcau(a,b):
    found = True
    dic = {}
    for i in range(len(a)):
        if a[i] not in dic:
           dic[a[i]] = b[i]
        else:
            if dic[a[i]] != b[i]:
                found = False
                break
    return found

dangcau(a,b)
            
```
