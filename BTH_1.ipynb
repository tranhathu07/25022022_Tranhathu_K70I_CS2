```python
n = 4              # Gán giá trị 4 cho biến n
t = 0              # Khởi tạo biến t bằng 0 (dùng để tích lũy kết quả)

for i in range(1, n+1):   # Vòng lặp for chạy i từ 1 đến n (bao gồm n)
    t += 1                # Mỗi lần lặp, tăng giá trị của t thêm 1

print(t)           # In ra giá trị cuối cùng của t
```

```python
number = {1, 2, 3, 4}           # Tạo một tập hợp (set) chứa các số 1, 2, 3, 4

for x in number:                # Duyệt qua từng phần tử x trong tập hợp number
    if x % 2 == 0:              # Nếu x chia hết cho 2 (số chẵn)
        print(x, 'là số chẵn')  # Thì in ra: x là số chẵn
    else:                       # Ngược lại (không chia hết cho 2 tức là số lẻ)
        print(x, 'là số lẻ')    # In ra: x là số lẻ
```

```python
animals = ['cat', 'dog', 'cat', 'bird']   # Tạo một danh sách gồm 4 phần tử
count = 0                                # Khởi tạo biến count = 0 (dùng để đếm)

for a in animals:                        # Duyệt qua từng phần tử a trong danh sách animals
    count += 1                           # Mỗi lần duyệt, tăng count thêm 1

print('Số phần tử trong danh sách là:', count)   # In ra kết quả cuối cùng
```
