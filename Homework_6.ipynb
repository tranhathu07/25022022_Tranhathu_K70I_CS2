```python
#W6A4

lst = list(input().split())
dic = {}
for i in lst:
    key,value = i.split(':')
    if key in dic:
        dic[key].append(value)
    else:
        dic[key] = [value]
print(dic)
```

```python
#W6A9
n = int(input())
a = {}
for _ in range(n):
    k,v = input().split(':')
    a[k] = int(v)

m = int(input())
b = {}
for _ in range(m):
    key,value = input().split(':')
    b[key] = int(value)

p = max(m,n)
c = {}
for k,v in a.items():
    c[k] = c.get(k,0)+v #Nếu có k thì thêm k còn ko thì thêm 0
for key,value in b.items():
    c[key] = c.get(key,0) + value
print(sorted(c.items(),key = lambda x:x[0]))
```

```python
#W6A10
lst = list(map(int,input().split()))
k = int(input())
ans= []
for i in range( len(lst)):
    for j in range(i+1,len(lst)):
        if lst[j] == k-lst[i]:
            ans.append((lst[i],lst[j]))
            
print(ans)
```

```python
#W6A12
lst = list(input().split())
dic = {}
maxa = 0
for i in lst:
    if i in dic:
        dic[i] += 1
        if dic[i]>maxa:
            maxa = dic[i]
    else:
        dic[i] = 1

sorted(dic.items(),key = lambda x: x[0])

for i in dic:
    if dic[i] == maxa:
        print(i)
        break

        
```
