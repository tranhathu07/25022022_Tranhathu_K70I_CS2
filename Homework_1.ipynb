```python
#Bài 6
import math
a,b,c = map(float, input('Nhập 3 số a,b,c: ').split())
if abs(a-b)<c<a+b and abs(a-c)<b< a+c and abs(b-c)<a<b+c:
    p = (a+b+c)/2
    s = math.sqrt(p*(p-a)*(p-b)*(p-c))
    print('Diện tích tam giác đó là',s)
else:
    print('Khong phai 3 canh cua tam giac')
```

```python
#Bài 8
import math
host = input('Ten chu ho:')
last = float(input('Chi so thang truoc:'))
now = float(input('Chi so thang nay:'))
def tiendien(x):
    tien = 0
    if 0<=x<=50:
        tien = 1984*x
    elif 51<=x<=100:
        tien = 1984*50 + 2050*(x-50)
    elif 101<=x<=200:
        tien = 1984*50+2050*50+2380*(x-100)
    elif 201<=x<=300:
        tien =1984*50+2050*50+2380*100 +2998*(x-200)
    elif 301<=x<=400:
        tien =1984*50+2050*50+2380*100 +2998*100 + 3350*(x-300)
    elif x>=401:
        tien =1984*50+2050*50+2380*100 +2998*100 + 3350*100 +3460*(x-400)
    return tien
print(f'Ho va ten:{host}\nTien phai tra là:{int(round(tiendien(abs(last-now))*1.08,0))}')
```
